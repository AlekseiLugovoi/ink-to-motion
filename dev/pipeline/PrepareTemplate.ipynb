```python
%load_ext autoreload
%autoreload 2

import os
import io
import cv2
import json
import numpy as np

from PIL import Image
from IPython.display import Image as IPImage, display
import matplotlib.pyplot as plt

from utils import (
    generate_template,
    overlay_character,
    template_meta,
    build_triangulation,
    draw_triangulation,
    markup_skeleton,
    save_skeleton,
    animate_character,
    render_svg,

    preview_motion,
    build_arap_mesh,
    draw_arap_mesh,
    animate_arap
)


# --- параметры шаблона ---
CANVAS_SIZE = (1240, 1754)   # A4 @ 150 DPI (h, w)
MARKER_PX = 80
MARGIN = 60
CONTENT_PAD = 10

TITLE = "SKETCH FACTORY"
OUTPUT_DIR = "output"

CHAR_ID = "004"
CHAR_DIR = f"templates/{CHAR_ID}"

meta = template_meta(CANVAS_SIZE, MARKER_PX, MARGIN, CONTENT_PAD)
print(f"Canvas:      {CANVAS_SIZE[1]}x{CANVAS_SIZE[0]}")
print(f"Drawing bbox: {meta['drawing_bbox']}")
print(f"Markers:      {meta['marker_centers']}")
```

### Общий Аквариум

```python
from PIL import Image, ImageOps
import numpy as np, cv2, json, io
from IPython.display import Image as IPImage, display
from utils import render_svg, animate_arap, DEFAULT_MOTION

FPS = 30
PREVIEW_SIZE = (900, 500)
pw, ph = PREVIEW_SIZE

# --- загрузка всех персонажей ---
def load_char(char_id):
    d = f"templates/{char_id}"
    with open(f"{d}/skeleton.json") as f:
        kp = json.load(f)["keypoints"]
    with open(f"{d}/animation.json") as f:
        anim = json.load(f)
    with open(f"{d}/motion.json") as f:
        patterns = json.load(f)
    img = cv2.cvtColor(render_svg(f"{d}/mask.svg"), cv2.COLOR_RGBA2BGRA)
    frames = animate_arap(img, kp, anim, fps=FPS, duration=3, pad=(50, 80, 0, 120))
    return frames, patterns

frames_001, patterns_001 = load_char("001")
frames_002, patterns_002 = load_char("002")
frames_003, patterns_003 = load_char("003")

# --- актёры ---
actors = [
    {"frames": frames_001, "patterns": patterns_001, "pat": "diag_up_left",  "scale": 0.30, "phase": 0.0},
    {"frames": frames_002, "patterns": patterns_002, "pat": "left",          "scale": 0.25, "phase": 0.3},
    {"frames": frames_003, "patterns": patterns_003, "pat": "up_center",     "scale": 0.20, "phase": 0.6},
]

def make_actor(a):
    cfg = dict(DEFAULT_MOTION)
    cfg.update(a["patterns"][a["pat"]])
    fr = a["frames"]

    direction = cfg["direction"]
    dir_rad = np.radians(direction)
    cos_d, sin_d = np.cos(dir_rad), np.sin(dir_rad)

    char_h = int(ph * a["scale"])
    char_w = int(char_h * fr[0].shape[1] / fr[0].shape[0])
    swim_dur = cfg["swim_duration"]

    span_x, span_y = pw * 0.35, ph * 0.35
    offset_x = cfg.get("offset_x", 0) / 100 * pw
    ax = pw / 2 - cos_d * span_x + offset_x
    ay = ph / 2 + sin_d * span_y
    bx = pw / 2 + cos_d * span_x + offset_x
    by = ph / 2 - sin_d * span_y

    bob_lo, bob_hi = cfg["bob_range"]
    bob_amp = ph * (bob_hi - bob_lo) / 200
    tilt_dur = cfg["tilt_duration"]
    base_angle = cfg.get("rotate", 0)
    do_flip = cfg.get("flip", False)

    turn_sign = 1 if not do_flip else -1
    turn_R = char_h * 0.3
    perp_angle = np.radians(direction + 90 * turn_sign)
    perp_sx = np.cos(perp_angle)
    perp_sy = -np.sin(perp_angle)

    turn_frac = 0.06
    fwd_end = 0.5 - turn_frac
    ret_start = 0.5 + turn_frac

    return dict(frames=fr, cfg=cfg, char_w=char_w, char_h=char_h, swim_dur=swim_dur,
                ax=ax, ay=ay, bx=bx, by=by, bob_amp=bob_amp, tilt_dur=tilt_dur,
                base_angle=base_angle, do_flip=do_flip, turn_sign=turn_sign,
                turn_R=turn_R, perp_sx=perp_sx, perp_sy=perp_sy,
                fwd_end=fwd_end, ret_start=ret_start, phase=a["phase"])

def draw_actor(a, t_global, canvas):
    t = (t_global + a["phase"]) % 1.0

    if t <= a["fwd_end"]:
        frac = t / a["fwd_end"]
        x = a["ax"] + (a["bx"] - a["ax"]) * frac
        y = a["ay"] + (a["by"] - a["ay"]) * frac
        phase = frac * a["swim_dur"] / a["tilt_dur"]
        y += a["bob_amp"] * np.sin(2 * np.pi * phase)
        tilt_v = a["cfg"]["tilt"] * np.sin(2 * np.pi * phase)
        angle = a["base_angle"] + tilt_v
    elif t <= a["ret_start"]:
        tp = (t - a["fwd_end"]) / (a["ret_start"] - a["fwd_end"])
        smooth = 0.5 - 0.5 * np.cos(np.pi * tp)
        angle = a["base_angle"] + a["turn_sign"] * 180 * smooth
        arc = a["turn_R"] * np.sin(np.pi * tp)
        x = a["bx"] + arc * a["perp_sx"]
        y = a["by"] + arc * a["perp_sy"]
    else:
        frac = (t - a["ret_start"]) / (1 - a["ret_start"])
        x = a["bx"] + (a["ax"] - a["bx"]) * frac
        y = a["by"] + (a["ay"] - a["by"]) * frac
        phase = frac * a["swim_dur"] / a["tilt_dur"]
        y += a["bob_amp"] * np.sin(2 * np.pi * phase)
        tilt_v = a["cfg"]["tilt"] * np.sin(2 * np.pi * phase)
        angle = a["base_angle"] + a["turn_sign"] * 180 + tilt_v

    fi = int(t_global * len(a["frames"])) % len(a["frames"])
    pil_char = Image.fromarray(cv2.cvtColor(a["frames"][fi], cv2.COLOR_BGRA2RGBA))
    if a["do_flip"]:
        pil_char = ImageOps.mirror(pil_char)
    pil_char = pil_char.resize((a["char_w"], a["char_h"]), Image.LANCZOS)
    pil_char = pil_char.rotate(angle, expand=True, resample=Image.BICUBIC,
                               fillcolor=(0, 0, 0, 0))
    cw, ch = pil_char.size
    canvas.paste(pil_char, (int(x) - cw // 2, int(y) - ch // 2), pil_char)

# --- рендер ---
actor_data = [make_actor(a) for a in actors]
total_dur = max(a["swim_dur"] for a in actor_data) * 2
n_frames = FPS * total_dur

preview = []
for i in range(n_frames):
    t = (i / n_frames) % 1.0
    canvas = Image.new("RGBA", (pw, ph), (100, 120, 140, 255))
    for a in actor_data:
        draw_actor(a, t, canvas)
    preview.append(canvas.convert("RGB"))

buf = io.BytesIO()
preview[0].save(buf, format="GIF", save_all=True, append_images=preview[1:],
                duration=1000 // FPS, loop=0)
print(f"Актёров: {len(actors)}, кадров: {n_frames}")
display(IPImage(data=buf.getvalue(), width=700))
```
